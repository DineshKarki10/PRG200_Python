# Filtering all ATM cash withdrawals 

```python
atm_withdrawals = df[(df["channel"] == "ATM") & (df["transaction_type"]=="Cash Withdrawal")]
print(atm_withdrawals.head())
print(f"ATM cash withdrawals: {len(atm_withdrawals)}")
```

```python
df.columns
```

```python
# All Failed, or reversed transactions
not_successful = df[df["transaction_status"] != "Success"]
print(f"Not Successful: {len(not_successful)} out of {len(df)} {len(not_successful)/len(df):.1%}")
not_successful["transaction_status"].value_counts()
```

```python
# High-value fund trsanfers
large_transfers = df[(df["transaction_type"]=="Fund Transfer") & (df["amount_npr"]>5000)]
print(f"Large fund Transfer(> NPR 50,000): {len(large_transfers)}")
```
